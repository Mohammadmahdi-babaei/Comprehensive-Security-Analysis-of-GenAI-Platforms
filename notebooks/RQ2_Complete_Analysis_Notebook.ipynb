# Generative AI / AI Agent vulnerability analysis - RQ2 (most frequent CWE types), 2023-2026
Everything below is recomputed from `2_GenAI_Vulnerabilities_Clean_Dataset_Step4.csv`; no number is typed by hand.

**Method notes**
1. Denominators: N = all unique CVEs in the clean sample; n_valid = CVEs with a specific CWE; NVD-CWE-noinfo records are reported separately.
2. One primary CWE per CVE (1-to-1 mapping). Cumulative share = cumulative raw count / n_valid, rounded once at the end.
3. Ranking: count desc, mean CVSS desc, CWE id asc. Ties at 2 CVEs are all listed (Top 5 + all recurring CWEs).

```python
import pandas as pd
df = pd.read_csv('2_GenAI_Vulnerabilities_Clean_Dataset_Step4.csv', keep_default_na=False)
valid = df[df['is_valid_cwe'].astype(str) == 'True']; noinfo = df[df['is_valid_cwe'].astype(str) == 'False']
print(f'Total verified records (N): {len(df)}')
print(f'Valid specific CWE (n_valid): {len(valid)} ({len(valid)/len(df)*100:.2f}%)')
print(f'NVD-CWE-noinfo (n_noinfo): {len(noinfo)} ({len(noinfo)/len(df)*100:.2f}%)')
print('Records by year:', df.groupby('published_year').size().to_dict())
```

```python
dist = pd.read_csv('3_RQ2_Top_Valid_CWE_Distribution.csv', keep_default_na=False)
print(dist[['rank','cwe','cwe_category','count','share_of_valid_cwe_pct','share_of_total_cve_pct','cumulative_valid_pct','mean_cvss']].to_string(index=False))
recurring = dist[dist['count'] >= 2]
print(f"\nTop 5 cover {dist['count'].head(5).sum()} of {len(valid)} valid CVEs ({dist['count'].head(5).sum()/len(valid)*100:.2f}%)")
print(f"All {len(recurring)} recurring CWEs (>=2 CVEs) cover {recurring['count'].sum()} of {len(valid)} ({recurring['count'].sum()/len(valid)*100:.2f}%)")
```
